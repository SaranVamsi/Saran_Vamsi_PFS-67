```python
students = {
    "Saran" : 80,
    "Vamsi" : 85,
    "Harsha" : 88
}

def add(name, marks):
    students[name] = marks
    return students

def view():
    return students

def search(name):
    return students[name]

def avg():
    return sum(students.values())/len(students)

def highest():
    d = 0
    for i,j in students.items():
        if j > d :
            d = j
    return d
    #return max(students[marks])

while True :
    print("1.Add Student")
    print("2.View Student")
    print("3.Search Student")
    print("4.Calculate Average")
    print("5.Find Highest Marks")
    print("6.Exit")
    
    ch = int(input("Enter your choice(1-6): "))

    if ch == 1 :
        a = input("Enter a student's name")
        b = int(input("Enter student's marks"))
        print("Adding Student")
        print(add(a,b))
    elif ch == 2 :
        print("2.View Student")
        print(view())
    elif ch == 3 :
        sn = input("Enter a name")
        print("3.Search Student")
        print(search(sn))
    elif ch == 4 :
        print("4.Calculate Average")
        print(avg())
    elif ch == 5 :
        print("5.Find Highest Marks")
        print(highest())
    elif ch == 6 :
        print("Exiting")
        break
    else : 
        print("Invalid Input")
```

# Random Password Generator

```python
import random
n = int(input("Enter range of password: "))
char = "qwertyuiopasdfghjklzxcvbnmQWERTYUIOPASDFGHJKLZXCVBNM0123456789!@#$%^&*<>;:"
password = ""
for i in range(n):
    password += random.choice(char)
print("password: ", password)
```

```python
import random
import string as s
char = s.ascii_letters + s.digits + s.punctuation
password = ""
for i in range(n):
    password += random.choice(char)
print("password: ", password)
```

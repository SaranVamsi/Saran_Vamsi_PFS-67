# OPERATORS

**Arithmetic**

```python
#Arithmetic :
a = 10
b = 3
print("add",a+b)
print("sub",a-b)
print("mul",a*b)
print("exp",a**b)
print("div",a/b)
print("floor",a//b)
print("mod",a%b)
```

**Relational**

```python
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
```

**Assignment**

```python
a+=b
print(a)
a-=b
print(a)
a*=b
print(a)
a**=b
print(a)

a/=b
print(a)

a//=b
print(a)

a%=b
print(a)

```

**Logical**

```python
x = 10
y = 3
print(x and y) # both are true. so, returns last value => 3
print(x or y) # returns the first true value => 10
print(not y) # 3 is true, so it prints False
```

**Identity**

```python
# is operator
a = 10
b = 10
print(a is b)
```

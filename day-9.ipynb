# LOOPS PRACTICE

**PALINDROME**

```python
n = input("Enter a number : ")
num = n[::-1]
if n == num :
    print("Palindrome")
else :
    print("Not palindrome")
```

```python
# 121, 12321, 1234321
n = int(input("Enter a number : "))
num = n
reverse = 0
while n > 0 :
    dig = num %10
    reverse = reverse * 10 +dig
    num //= 10
if num == reverse :
    print("Palindrome")
else :
    print("Not a palindrome")
```

**FACTORIAL**

```python
n = int(input("Enter a number : "))
num = n
fact = 1
while num > 0 :
    fact *= num
    num -= 1
print(f"Factorial of {n} : {fact}")
```

**SUM OF DIGITS**

```python
#1234
n = int(input("Enter a number : "))
sum = 0
while n > 0 :
    dig = n % 10 #4
    sum += dig
    n //= 10
print(sum)
```

**Fibonacci**

```python
#011235813
n = int(input("Enter range of series : "))
first = 0
nxt = 1
while n > 0 :
    print(first)
    fib = first + nxt
    first = nxt
    nxt = fib
    n -= 1
print(fib)
```

```python
n = int(input("Enter range of series : "))
a = 0
b = 1
for i in range(n) :
    print(a, end = " ")
    a, b = b, a+b
```

**LARGEST DIGIT IN NUM**

```python
n = int(input("Enter a number : "))
l = 0
while n > 0 :
    dig = n % 10
    if dig > l :
        l = dig
    n = n // 10
print(l)
```

**Smallest Digit**

```python
n = int(input("Enter a number : "))
s = 9
while n > 0 :
    dig = n % 10
    if dig < s :
        s = dig
    n = n // 10
print(s)
```

**Duplicate numbers**

```python
n = [1,2,3,4,5,2,6,7,8,9,1]
for i in range(len(n)):
    for j in range(i+1, len(n)):
        if n[i] == n[j] :
            print(n[i])
```

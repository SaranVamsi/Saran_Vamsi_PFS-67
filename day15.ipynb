# GENERATORS

```python
def num():
    yield 1
    yield 2
    yield 3
res = num()
print(next(res))
print(next(res))
print(next(res))
```

```python
def num():
    for i in range(1,6):
        yield i
res = num()
print(next(res))
print(next(res))
print(next(res))
print(next(res))
print(next(res))
```

```python
def num():
    for i in range(1,6):
        yield i
res = num()
for i in res :
    print(i)
```

**yield vs return**

```python
def demo():
    return 10
    return 20
print(demo())
print(demo())
```

```python
def demo():
    yield 10
    yield 20
    yield 30
res = demo()
for i in res :
    print(i)
```

```python
def demo():
    print("start")
    yield 10
    print("Stop")
res = demo()
for i in res :
    print(i)
```

```python
def demo():
    print("start")
    yield 10
    print("Stop")
res = demo()
print(next(res))

```

```python
n = (i * i for i in range(1,6))
print(n)
```

```python
n = (i * i for i in range(1,6))
print(next(n))
print(next(n))
print(next(n))
print(next(n))
print(next(n))
```

```python
n = (i * i for i in range(1,6))
for i in n :
    print(i)
```

```python
def even():
    for i in range(1,21):
        if i % 2 == 0 :
            yield i
res = even()
for i in res :
    print(i)
```

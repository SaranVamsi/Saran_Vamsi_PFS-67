# INHERITENCE

**1.Single**

```python
class animal:
    def eat(self):
        print("Animal eats")
class lion(animal):
    def roar(self):
        print("Lion roar")
o = lion()
o.eat()
o.roar()
```

**2.Multiple**

```python
class Father:
    def fprop(self):
        print("Mom gives house")
class Mother :
    def mprop(self):
        print("Dad gives Land")
class child(Father, Mother):
    def childprop(self):
        print("Child properties")
o = child()
o.fprop()
o.mprop()
o.childprop()
```

**3.Multi - Level**

```python
class grandpa:
    def house(self):
        print("Grandpa gave house")
class Father(grandpa):
    def land(self):
        print("Dad gave land")
class child(Father):
    def bike(self):
        print("I bought a bike")
o = child()
o.house()
o.land()
o.bike()
```

**4.Hierarchial**

```python
class animal:
    def eat(self):
        print("Animal eats")
class lion(animal):
    def roar(self):
        print("Lion roar")
class cat(animal):
    def meow(self):
        print("Meow")
a = lion()
b = cat()
a.eat()
a.roar()
b.eat()
b.meow()
```

**5.Hybrid**

```python
class A :
    def a(self):
        print("A")
class B(A) :
    def b(self):
        print("B")
class C(A) :
    def c(self):
        print("C")
class D(B,C) :
    def d(self):
        print("D")
o = D()
o.a()
o.b()
o.c()
o.d()
```

# SUPER METHOD

```python
class parent:
    def show(self):
        print("Animal eats")
class child(parent):
    def sat(self):
        super().show()
        print("Lion roar")
o = child()
o.sat()
```

**Super with constructor**

```python
class Person:
    def __init__(self, name):
        self.name = name
class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks
s = Student("Saran", 80)
print(s.name)
print(s.marks)
```

```python
class Person :
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def disp(self):
        print(self.name)
        print(self.age)
class Student(Person):
    def __init__(self, roll, branch):
        self.roll = roll
        self.branch = branch
    def disp_std(self):
        print(self.roll)
        print(self.branch)
p = Student("saran",22,12,"hyd")
p.disp()
p.disp_std()
```

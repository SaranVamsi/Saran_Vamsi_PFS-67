# LIST

```python
List :
-> Collection of diff data types
-> Slicing 
-> Concatination
-> Ordered
-> Allow Duplicates
-> Mutable

List Methods :

1) append() adds an element to the end.
2) extend() appends elements from an iterable.
3) insert() places an element at a specific index.
4) remove() deletes the first matching value.
5) pop() removes and returns an item by index.
6) clear() empties the entire list.
7) index() finds the position of a value.
8) count() tallies occurrences of a value.
9) reverse() modify the sequence in-place.
10) copy() creates a shallow duplicate. 
11) len()
12) max()
13) min()
14) sum()
15) sort() - in ascending order.
    list.sort(reverse = True) sorts list in descending order.
```

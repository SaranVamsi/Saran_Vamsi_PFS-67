**Star Pattern**

# 1st

```python
for i in range(1, 6) :
    for j in range(i) :
        print("*", end=" ")
    print()
'''
When we use only print statement, 2nd iteration will be printed in next line.
'''
```

```python
for i in range(1, 6) :
    for j in range(i) :
        print("*", end=" ")
    print("\n", end="")
'''
\n with end="" means same as before.  
'''
```

```python
for i in range(1, 6) :
    for j in range(i) :
        print("*", end=" ")
    print("\n")
'''
Only \n means next 1 line gap.
'''
```

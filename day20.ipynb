**Encapsulation**

```python
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
    def withdraw(self,amount):
        self.balance -= amount
    def get_balance(self):
        return self.balance 
ba = BankAccount(50000)
ba.deposit(10000)
ba.withdraw(5000)
print(ba.get_balance())
  
```

```python
class Std :
    def __init__(self,marks):
        self.__marks = marks
    def get_marks(self):
        return self.marks
    def set_marks(self,marks):
        if 0 <= marks <= 100 :
            self.__marks = marks
        else :
            print("Invalid marks")
s = Std(80)
print(s.get_marks())
s.set_marks(90)
print("Moded: ",s.get_marks())
```

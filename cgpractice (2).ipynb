```python
def play():
    c1=0
    c2=0
    rock = ["rock", "Rock", "ROCK"]
    paper = ["paper", "Paper", "PAPER"]
    scissor = ["scissor", "Scissor", "SCISSOR"]
    n = int(input("No of rounds: "))
    for i in range(n):
        p1 = input("Player 1 enter: ")
        p2 = input("Player 2 enter: ")
        if p1 in rock :
            if p2 in scissor :
                #print("Player 1 is winner")
                c1 += 1
            elif p2 in paper :
                #print("Player 2 is winner")
                c2 += 1
            #else : 
                #print("It's a draw")
        elif p1 in paper :
            if p2 in rock :
                #print("Player 1 is winner")
                c1 += 1
            elif p2 in scissor :
                #print("Player 2 is winner")
                c2 += 1
        elif p1 in scissor :
            if p2 in paper :
                #print("Player 1 is winner")
                c1 += 1
            elif p2 in rock :
                #print("Player 2 is winner")
                c2 += 1
    if c1 > c2 :
        print("P1 wins")
    elif c2 > c1 :
        print("P2 wins")
    elif c1 == c2 :
        print("Draw")
play()
```

# Temperature Converter
# Celsius → Fahrenheit
# Fahrenheit → Celsius

```python
print("Temperature Converter")
def ctof(c):
    f = (c * 9/5) + 32
    return f
def ftoc(f):
    c = (f - 32) * 5/9
    return c
print("1. Celcius -> Fahrenheit")
print("2. Fahrenheit -> Celcius")
ch = int(input("Enter your choice(1 or 2): "))
if ch == 1 :
    a = int(input("Enter Celcius: "))
    print("F: ", ctof(a))
elif ch == 2 :
    b = int(input("Enter Fahrenheit: "))
    print("F: ", ftoc(b))
else :
    print("Invalid Choice")
```

# Number Analyzer

```python
'''
Number of digits
Sum of digits
Largest digit
Smallest digit
Reverse of number
'''
```

```python
def nod():
    temp = n
    no_of_dig = 0
    while temp > 0 : # 123
        dig = temp % 10 # 3
        no_of_dig += 1
        temp //= 10
    print("No of Digits: ", no_of_dig)

def sod():
    temp = n
    sum = 0
    while temp > 0 :
        dig = temp % 10
        sum += dig
        temp //= 10
    print("Sum of Digits: ",sum)

def ld():
    a = list(str(n))
    p = int(a[0])
    for i in a:
        if int(i) > p:
            p = int(i)
    print("Largest dig:", p)

def sd():
    a = list(str(n))
    p = int(a[0])
    for i in a:
        if p > int(i):
            p = int(i)
    print("Smallest dig:", p)

def ron():
    temp2 = n
    rev = 0
    while temp2 > 0 : #456
        dig2 = temp2 % 10 
        rev = rev * 10 + dig2
        temp2 //= 10
    print("Reverse: ",rev)

print("Number Analyser")
n = int(input("Enter a number: "))
print("1.No of Digits" )
print("2.Sum of Digits" )
print("3.Largest Digit" )
print("4.Smallest Digit" )
print("5.Reverse of the Number" )
c = int(input("Enter your choice(1-5): "))
if c == 1 :
    nod()
elif c == 2 :
    sod()
elif c == 3 :
    ld()
elif c == 4 :
    sd()
elif c == 5 :
    ron()
else :
    print("Invalid Choice")
    
```

# STUDENT MARKS

```python
students = {
    "Saran" : 80,
    "Vamsi" : 85,
    "Harsha" : 88
}

def add(name, marks):
    students[name] = marks
    return students

def view():
    return students

def search(name):
    return students[name]

def avg():
    return sum(students.values())/len(students)

def highest():
    d = 0
    for i,j in students.items():
        if j > d :
            d = j
    return d
    #return max(students[marks])

while True :
    print("1.Add Student")
    print("2.View Student")
    print("3.Search Student")
    print("4.Calculate Average")
    print("5.Find Highest Marks")
    print("6.Exit")
    
    ch = int(input("Enter your choice(1-6): "))

    if ch == 1 :
        a = input("Enter a student's name")
        b = int(input("Enter student's marks"))
        print("Adding Student")
        print(add(a,b))
    elif ch == 2 :
        print("2.View Student")
        print(view())
    elif ch == 3 :
        sn = input("Enter a name")
        print("3.Search Student")
        print(search(sn))
    elif ch == 4 :
        print("4.Calculate Average")
        print(avg())
    elif ch == 5 :
        print("5.Find Highest Marks")
        print(highest())
    elif ch == 6 :
        print("Exiting")
        break
    else : 
        print("Invalid Input")
```

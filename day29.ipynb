# Data Visulaization - Graphical representation of data

```python
import matplotlib.pyplot as m
```

**Line PLot - used to show trends or changes over time**

```python
months = ["J", "F", "M", "A", "S", "O"]
sales = [100, 120, 130, 140, 110, 160]
m.plot(months, sales)
m.xlabel("Month")
m.ylabel("Sales")
m.title("Monthly Sales")
m.show()
```

```python
months = ["J", "F", "M", "A", "S", "O"]
sales = [100, 120, 130, 140, 110, 160]
m.bar(months, sales)
m.xlabel("Month")
m.ylabel("Sales")
m.title("Monthly Sales")
m.show()
```

**Pie Chart - Shows how total is divided into diff categories**

```python
a=["Python", "Java", "AI"]
val = [40, 30, 30]
m.pie(val, labels = a, autopct = "%1.1f%%")
m.title("Course Distribution")
m.show
```

**Histogram - Shows distribution of numerical data**

```python
a = [35,40,42,54,65,78,95,90,100,85,80,75,45,62,18,97,79]
m.hist(a,bins = 5)
m.xlabel("Marks")
m.ylabel("No.of Students")
m.title("Marks Distribution")
m.show()
```

**Scatter Plot - Shows relationship between two numerical variables**

```python
hour = [1,2,3,4,5,6,7]
marks = [40,50,60,70,60,50,40]
m.scatter(hour, marks)
m.xlabel("Study Hours")
m.ylabel("Marks")
m.title("Marks vs Study Hours")
m.show()
```

**Box Plot - Use it to understand distribution, median, spread, outliners**

```python
hour = [1,2,3,4,5,6,7]
marks = [40,50,60,70,60,50,40]
m.boxplot(hour, marks)
m.ylabel("Marks")
m.title("Marks Distribution")
m.show()
```

**Heatmap - Represent values using different intensities/shades, making pattern easy to identify**

```python
import seaborn as s
d = {
    "Age" : [20,25,30,35,40],
    "Salary" : [20000,25000,30000,10000,15000],
    "Exp" : [1,3,5,7,9]
}
import pandas as pd
df = pd.DataFrame(d)
correlation = df.corr()
s.heatmap(correlation, annot=True)
m.show()
```

# SCOPE

**1) LOCAL SCOPE**

```python
def add():
    a = int(input())
    b = int(input())
    return a + b
add()
#print(a) - gives error. a is local variable, but is being executed globally.
```

```python
def a():
    x = 10
    print(x)
def b():
    x = 20
    print(x)
a()
b()
```

**2)GLOBAL SCOPE**

```python
a = int(input())#global scope
def add():
    b = int(input())#local scope
    return a + b
print(add())
print(a)
```

```python
a = int(input())#global scope
def add():
    a = 30
    b = int(input())#local scope
    return a + b
print("Local a value: ",a,"+",b,"=",add())
print("Global a value",a)
```

```python
x = 100
def func():
    global x
    x = 200
print(x)
func()
print(x)
'''
When you keep global x, global variable will be printed until the func is 
called. After the func is called, only local variable is printed for print(x).
'''
```

# PASS BY VAL AND PASS BY REFERENCE

```python
def disp(x):
    x = 20
    print(x)
a = 10
print(a)
disp(a)
print(a)
```

```python
def add(x):
    x += 10 #ref value
    print("Inside:", x)
n = 50 #original value
add(n) #changed reference value
print("Outside:", n) #original value. it wont change
```

**Pass by reference:**

```python
'''
Here, original val should also change
'''
def add(x):
    x.append(40)
a = [10,20,30]
print(a) #original val printed.
add(a) #function is called. original val changed.
print(a) #changed original val is printed.
```

# RECURSION

```python
'''
Base case -> The condition where the code execution should stop.
'''

def count(n):
    if n == 0 : #base case.
        return
    print(n)
    count(n-1)
x = 5
count(x)
```

```python
def fact(n):
    if n <= 1 :
        return 1
    return n * fact(n-1)
print(fact(5))
```

```python
def fib(n):
    if n <= 1 :
        return n
    
    return fib(n-1) + fib(n-2)
n = 5
for i in range(n):
    print(fib(i), end=" ")
```

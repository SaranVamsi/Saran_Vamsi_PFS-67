# Exception Handling


```python
try :
    a = int(input())
    b = int(input())
    res = a/b
except ZeroDivisionError :
    print("Cannot Divide a number by 0")
except ValueError:
    print("Please enter a valid value")
else :
    print(res)
finally :
    print("End")
```

```python
age = int(input())
if age < 0 :
    raise ValueError("Age cannot be negetive")
print("Age",age)
except ValueError as e :
    print("Error",e)
```

# FILE HANDLING

**Creating and Writing to a file**

```python
'''
"w" : rewrites the existing data in file
'''
file = open("data.txt", "w")

file.write("Hello Saran\n")
file.write("How are you?\n")
file.write("I'm doing amazing")

file.close()
```

**Reading a file**

```python
file = open("data.txt", "r")
data = file.read()
print(data)
file.close()
```

```python
file = open("data.txt", "r")

print(file.readline()) #readline() - only reads a single line

file.close()
```

**Append - Add data to existing file**

```python
file = open("data.txt","a")
file.write("\nVikas sir teaches python")
file.close()
```

**WithOpen()**

```python
with open("data.txt","r") as file :
    data = file.read()
    print(data)
```

```python
with open("data.txt","a") as file :
    file.write("\nPrajwal sir teaches attitute\n")
    file.write("Praveen sir teaches softskills")
```

```python
with open("data.txt","r") as file :
    for i in file :
        print(i.strip())
```

```python
name = input("Name of student: ")
age = input("Age of student: ")
course = input("Course of student: ")
with open("std.txt", "w") as file :
    file.write(f"Name: {name}")
    file.write(f"Age: {age}")
    file.write(f"Course: {course}")
print("Students details saved successfully")

with open("std.txt", "r") as file :
    print(file.read())    
```

```python
balance = 10000
try :
    amt = int(input())
    if amt <= 0 :
        raise ValueError("Amount should be more than 0")
    if amt > balance :
        print("Insufficient Balance")
    else :
        balance -= amt
        with open("std2.txt", "w") as file :
            file.write(f"Withdrawn Amt : {amt}\n")
            file.write(f"Balance: {balance}\n")
        print("Success")
        print("Balance: ", balance)
except ValueError as e :
    print("Error: ", e)
```

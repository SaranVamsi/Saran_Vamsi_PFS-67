# STRING SLICING

```python
a = "codegnan"
print("From index 0 to 5 : ", a[0:5])
print("From index 0 to 5 : ", a[:5])
print("From index 3 to ending : ", a[3:])
print("From index 3 to -1 : ", a[:-1])
print("From index 0 to -2 : ", a[:-2])
print("From index 3 to ending : ", a[-1:])


```

```python
a = "python"
print(a[0:6:1]) # Here, 1 means the 1st next number to be printed after index 0
print(a[0:6:2]) # Here, 2 means the 2nd next number to be printed after index 0
print(a[::-1])
print(a[::-2])
print(a[-6:-1]) # o/p : pytho. Index is -1, so only till -2 is printed
print(a[-6:0]) 
'''Nothing is printed casue 0 is neither odd nor positive. So, when 0 is 
   placed in the right side of index, machine is confused and prints nothing'''
print(a[5:0]) # no o/p


```

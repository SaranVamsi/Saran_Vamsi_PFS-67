**5.Count of even and odd**

```python
def coun(n, e, o):
    for i in n :
        if i % 2 == 0 :
            e += 1
        elif i % 2 != 0 :
            o += 1
    return "Even: ",e, "Odd: ",o
    
a = list(map(int, input().split()))
b = 0
c = 0
print(coun(a,b,c))
```

**6.Remove Duplicates**

```python
def dup(n):
    new = []
    for i in n :
        if i not in new :
            new.append(i)
    return new
a = list(map(int, input().split()))
print(dup(a))
```

**7.Reverse a list**

**13.Find the sum of digits of every element**

```python
a = list(map(int, input().split()))
n = 0
for i in a :
    while i > 0 :
        dig = i % 10
        n += dig
        i //= 10
print(n)
```

**17.Print dupes in a list**

```python
a = list(map(int, input().split()))
dupe = []
for i in range(len(a)):
    for j in range(i+1, len(a)):
        if a[i] == a[j]:
            dupe.append(a[i])
print(dupe)
        
```

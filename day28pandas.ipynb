# DataFrame() Data Structure

```python
import pandas as pd
d = {
    "name" : ["A","B","C"],
    "age" : [20,22,21],
    "marks" : [80,90,70]
}
df = pd.DataFrame(d)
print(df)
```

**Accessing Columns**

```python
print(df["name"])
```

**Accessing Multiple Columns**

```python
print(df[["name","marks"]])
```

**Accessing Rows**

```python
print(df.iloc[0])
```

```python
print(df.iloc[1])
```

**Multiple Rows**

```python
print(df.iloc[0:2])
```

```python
import pandas as pd

d = {
    "name": ["A", "B", "C"],
    "age": [20, 22, 21],
    "marks": [80, 90, 70]
}

df = pd.DataFrame(d)


print(df.loc[0:1, ["name", "marks"]])
```

**Accessing specific value**

```python
print(df.loc[2,"marks"])
```

**Rows and columns togeather**

```python
print(df.iloc[0:2, 1:3])
```

**Adding a new column**

```python
import pandas as pd

d = {
    "name": ["A", "B", "C"],
    "age": [20, 22, 21],
    "marks": [80, 90, 70]
}

df = pd.DataFrame(d)

df["Pass"] = True
print(df)
```

```python
df["Pass"] = df["marks"] >70
print(df)
```

**Adding column with values**

```python
df["City"] = ["Hyderabad", "Delhi", "Chennai"]
print(df)
```

**Removing a column**

```python
import pandas as pd

d = {
    "name": ["A", "B", "C"],
    "age": [20, 22, 21],
    "marks": [80, 90, 70]
}
df = pd.DataFrame(d)
df["City"] = ["Hyderabad", "Delhi", "Chennai"]
print(df)
df = df.drop("City", axis=1)
print(df)
```

```python
import pandas as pd
df = pd.read_csv("/kaggle/input/datasets/saranvamsipva/code-csv/data (1).csv")

df["Calories"] = df["Calories"].fillna("2026")
print(df)
```

```python
df["Calories"] = df["Calories"].fillna(df["Calories"].mean)
print(df)
```

**Print duplicates**

```python
print(df.duplicated())
```

**Remove Duplicates**

```python
df = df.drop_duplicates()
print(df)
```

**Changing Column Names**

```python
df = df.rename(columns={
    "Date" : "date"
})
print(df.head())
```

```python
res = df[df["Duration"]>45]
print(res)
```

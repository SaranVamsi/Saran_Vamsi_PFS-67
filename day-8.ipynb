# 1)For Loop

**1)for loop using index**

```python
for i in range(1, 11):
    print("Hello")

'''
iteration-1 : i = 1
i2 : i = 2
..........
..........
i10 : i = 10
'''
```

```python
a = [10,20,30,40,50]
for i in range(0, len(a)):
    print(a[i])
```

**2)By using value**

```python
nums = [10,20,30,40,50]
for num in nums :
    print(num)
```

# 2)WHILE LOOP

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

# PRACTICE

**1)Even Numbers**

```python
i = 1
while i <= 10:
    if i % 2 == 0 :
        print(i)
    i += 1
```

```python
i = 10
while i <= 17:
    if i % 2 == 0 :
        print(i)
    i -= 1
```

```python
for i in range(10, 0, -2) :
    print(i)
```

**2)Sum of numbers**

```python
total = 0
for i in range(1,11) :
    total+=i
print(total)
```

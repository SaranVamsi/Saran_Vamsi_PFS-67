# SET FUNCTIONS

```python
1) set.discard() : removes ele but dosen't give error if ele dosen't exist.
2) set1.union(set2) : common eles are only displayed once.
3) set1.intersection(set2) : displays only common elements
4) set1.difference(set2) : displays set1 eles which are not commom with set2.
5) set1.symmetric_difference(set2) : both sets eles which are not commom.
6) set1.issubset(set2) : T/F
   a = {1, 2}
   b = {1, 2, 3}
7) set1.issuperset(set2) : T/F
   a = {1, 2, 3}
   b = {1, 2}
8) set1.isdisjoint(set2) : [T/F] if both are not same
   
```

```python
s = {10, 20,30}
s.add(40)
print(s)
```

```python
#Union
a = {10,20,30,40}
b = {30,40,50,60}
print(a|b) # or print(a.union(b))
```

```python
#Intersection
print(a&b)# or print(a.intersection(b))
```

```python
# Difference
print(a-b) # or print(a.difference(b))
```

```python
#Symmetric_difference
print(a^b)
```

# ABSTRACTION

```python
from abc import ABC as a, abstractmethod as am
class Animal(a):
    @am
    def sound(self):
        pass
#Animal says that every animal must provide a sound() method
class  Dog(Animal):
    def sound(self):
        print("Bark")
       
class Cat(Animal):
    def sound(self):
        print("Meow")
d = Dog()
d.sound()
c = Cat()
c.sound()
```

```python
from abc import ABC as a, abstractmethod as am
class Payment(a):
    @am
    def pay(self,amt):
        pass
class UPI(Payment):
    def pay(self, amt):
        print("Paid", amt, "using UPI")
class Credit(Payment):
    def pay(self, amt):
        print("paid", amt, "usind card")
u = UPI()
c = Credit()
u.pay(5000)
c.pay(45000)
```

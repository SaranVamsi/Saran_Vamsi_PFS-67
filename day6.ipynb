**ATM Example**

```python

balance = int(input("Balance: "))
print("Please Insert Your Card")
card = input("Enter your card number: ")
if card == "12345678" :
    print("Welcome")
    print("Enter your ATM Pin")
    pin = int(input("Please enter your pin :"))
    if pin == 3699 :
        print("Your current balance: ", balance)
        withdrawl_amt = int(input("Enter Withdrawl Amount : "))
        if balance >= 1000 and withdrawl_amt <= balance:
            if balance %10 == 0 :
                print("Amount Withdrawn :", withdrawl_amt)
                print("Transaction Successfull")
                print("Please collect your card")
            else :
                print("Please enter digits which are multiples of 10")
        else :
            print("Your Balance is insufficient :", balance)
    else :
        print("Invalid pin")
        print("Enter the correct pin")
else :
    print("Card is Invalid")
    print("Please check your card")
```

**Online Shopping Example**

```python
'''Product available
login account
is payment done'''

product = input("Is product available? (Yes/No): ")
if product == "Yes" :
    print("The product is available")
    account = input("Enter your account name: ")
    if account == "saran@gmail" :
        print("You are logged in")
        payment = input("Is your payment done? (Yes/No): ")
        if payment == "Yes" :
            print("Payment successfull")
            print("Your order will be delivered soon.")
            print("Please visit again.")
        else :
            print("Payment failure. Please try again.")
    else :
        print("Login failed. Please try again.")
else:
    print("Sorry the product is unavailabe right now.")
    print("Please visit the product again")
```

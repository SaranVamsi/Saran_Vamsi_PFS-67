**re.findall()**

```python
#It finds all occurances of a pattern and returns it as a list
import re
txt = "I have 10 applese and 20 oranges"
res = re.findall(r"\d+", txt) 
#\d - digits from (0-9)
#\d+ - combo of 1 or more digits
print(res)
print(type(res))
```

**^ - star of string**

```python
#The pattern must start at beginning
import re
a = r"^Python"
print(re.search(a, "Python is easy"))
```

**$ - Dollar**

```python
#Pattern must end at the end of string
import re
a = r"Python$"
print(re.search(a, "is easy Python"))
```

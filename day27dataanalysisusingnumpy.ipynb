**2D Slicing**

```python
import numpy as np
m = np.array([
    [80, 90, 70],
    [85, 95, 75],
    [70, 80, 90]
])
print(m[0:2,1:3])
#[0:2] -> row 0 and 1
#[1:3] -> columns 1 and 2
```

# Array Operations

**Arithmetic**

```python
a = np.array([10,20,30])
b = np.array([1,2,3])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
```

**2)Comparision**

```python
#Boolean Values
a = np.array([10,20,30])
print(a>20)
```

```python
#Values
a = np.array([10,20,30])
print(a[a>20])
```

```python
print(a==20)
```

```python
print(a[a != 20])
```

```python
print(a[a>=20])
```

# Mathematical Functions

```python
m = np.array([75,80,85,90,95])
print(np.sum(m))
print(np.max(m))
print(np.min(m))
print(np.mean(m))#Average
```

```python
m = np.array([
    [80, 90, 70],
    [85, 95, 75],
    [70, 80, 90]
])

#Colum - wise sum
print("Column wise: ",np.sum(m, axis=0))

#Row - wise sum
print("Row wise: ",np.sum(m, axis=1))
```
